## Sentiment Analysis using pipeline

```python
from transformers import pipeline

p = pipeline("sentiment-analysis",model="unitary/toxic-bert")
p("I really enjoyed this movie!")
```

## Zero Shot Classification

```python
zs = pipeline("zzero-shot-classification")
```

```python
zs("The movie is good", candidate_labels=["Movie","Sentiment","Politics"])
```

## Fill Mask Pipeline

```python
c=pipeline('fill-mask',model='google-bert/bert-base-uncased')

result = c("The movie is [MASK].")
print(result)
```

## Named entity recognition

```python
c=pipeline('ner')

c("Rajini is a good actor in Tamilnadu")
```

## Question Answering Pipeline

```python
c=pipeline('question-answering')
result = c(question="where do i work",context="i am working at vit chennai")
print(result)
```

## Summarization Pipeline

```python
c=pipeline('summarization')
result = c("It was established under Section 3 of the University Grants Commission (UGC) Act, 1956, and was founded in 1984 as a self-financing institution called the Vellore Engineering College. The Union Ministry of Human Resources Development conferred University status on Vellore Engineering College in 2001. The University is headed by its founder and Chancellor, Dr. G. Viswanathan, a former Parliamentarian and Minister in the Tamil Nadu Government. In recognition of his service to India in offering world class education, he was conferred an honorary doctorate by the West Virginia University, USA.VIT Chennai was established in 2010. Mr. Sankar Viswanathan, Dr. Sekar Viswanathan and Mr.G.V.Selvam are the Vice-Presidents; Dr. V. S. Kanchana Bhaaskaran is the Vice-Chancellor.")
print(result)
```
